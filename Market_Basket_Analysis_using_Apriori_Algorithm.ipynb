```python
transactions=[]
for i in range(0,len(df)):
  transactions.append([str(df.values[i,j]) for j in range(0,20) if str(df.values[i,j])!='nan'])
```

```python
from apyori import apriori
rules=apriori(transactions=transactions,min_support=0.003,min_confidence=0.2,min_lift=3,min_length=2,max_length=2)
results=list(rules)
results
```

```python
import pandas as pd

rows = []

for rule in results:
    support = rule.support

    for stat in rule.ordered_statistics:
        rows.append([
            ', '.join(list(stat.items_base)),      # Antecedent
            ', '.join(list(stat.items_add)),       # Consequent
            round(support, 4),
            round(stat.confidence, 4),
            round(stat.lift, 4)
        ])

df_rules = pd.DataFrame(
    rows,
    columns=['lhs', 'rhs', 'Support', 'Confidence', 'Lift']
)

# Sort by Lift (highest first)
df_rules = df_rules.sort_values(by='Lift', ascending=False)

df_rules
```
